## Checking equations of the flows

```julia
uj=rand(4)
uj1=copy(uj)
flowH1Schwarzschildv3!(uj1,[],0.1,parms)
@show(norm(uj1-uj))
@show H1v3(uj1,parms)/H1v3(uj,parms)-1.;
```

```julia
uj=rand(4)
uj1=copy(uj)
flowH2Schwarzschildv3!(uj1,[],0.1,parms)
@show(norm(uj1-uj))
@show H2v3(uj1,parms)/H2v3(uj,parms)-1.;
```

```julia
uj=rand(4)
uj1=copy(uj)
flowH3Schwarzschildv3!(uj1,[],0.1,parms)
@show(norm(uj1-uj))
@show H3v3(uj1,parms)/H3(uj,parms)-1.;
```
